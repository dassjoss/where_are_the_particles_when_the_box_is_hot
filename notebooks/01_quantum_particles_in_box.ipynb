## Introducción

En el marco de la mecánica cuántica, se han desarrollado múltiples teorías que nos han permitido entender y simular sistemas de forma ideal. Permitiéndonos conocer el mundo microscópico de una forma más precisa.

Se desarrollan sistemas tales como la partícula en una caja, en un pozo potencial, sistemas de N cuerpos etc... En particular, un problema que permite observar las bases de la teoría de la mecánica cuántica es el problema de la partícula confinada en una caja. Este, aunque es un sistema idealista, proporciona una visión general de cómo se afrontan los problemas de forma genérica dentro de este marco.

Actualmente, existe una bibliografía abundante de este sistema clásico, pues permite un desarrollo matemático simple. Sin embargo, para acercarse más a la realidad, sobre el sistema se realizan ciertas variaciones, como incluir el spin, temperatura, o más partículas. Sin embargo, entre más características son agregadas que permiten una mejor similitud con la realidad más complejo es el problema al ser resuelto.

La característica que más se omite en estos sistemas, es la temperatura, esto debido a que la mecánica cuántica tiene dificultades a la hora de enmarcar esta característica física dentro de la teoría. Sin embargo, la física estadística permite la incorporación de la temperatura dentro de estos sistemas, dando una mayor aproximación a lo que sucede en la naturaleza.

En este sistema, como en todos los sistemas cuánticos, se busca conocer las características del mismo, la más usual de todas es la posición. Sin embargo, como esta no se puede conocer de forma exacta, se habla de la probabilidad de encontrar al sistema en un intervalo específico. Por tanto lo que caracteriza al sistema en términos de la posición, es la densidad de probabilidad de encontrar al sistema en un intervalo específico.

## Marco teórico

En mecánica cuántica, se habla sobre la probabilidad de encontrar una partícula en una región del espacio específica en un tiempo determinado. Existe una función que permite describir al sistema(y las características mencionadas), esta función es la función de onda $ψ(x,t)$. En general esta función es compleja. Uno de los postulados de la mecánica cuántica es que la densidad de probabilidad de encontrar a el sistema en una posición y tiempo determinados viene dada por $P(x,t)=|Ψ(x,t)|^2$. Esta función debe de cumplir con la ecuación de Schrödinger

$$ i\hbar \frac{\partial}{\partial t} \Psi(\mathbf{r},t) = \hat{H} \Psi(\mathbf{r},t) $$

Cuando los estados a tratar son autoestados de $H$, es decir que cumplan con $ \hat{H} \Psi_{n} = E_{n} \Psi_{n} $ su solución es trivial

$$ \Psi_n(x, t) = \varphi_n(x) e^{-i E_n t / \hbar} $$


Donde $\Psi_{n}$ representa el autoestado n-ésimo y $E_{n}$ la energía asociada a ese autoestado, y $φ_{n}(x)$ una función asociada solo a la posición. De hecho en el caso simplificado de una partícula encerrada en una caja, de lado L, se supone un potencial infinito en las paredes de la misma, produciendo así condiciones de contorno, haciendo que la solución de cualquier función tenga una forma sinusoidal. Estas parte espaciales de la autofunción, y de la energía son discreto, más aún se conocen y son fáciles de encontrar

$$ E_n = \frac{h^2 n^2}{8m L^2}, \qquad \varphi_n(X) = \sqrt{\frac{2}{L}} \sin \left( \frac{n \pi}{L} X \right) $$

Así, la probabilidad de encontrar al autoestado n-ésimo denotada por $P_{n}$ tendrá la forma


$$ P_n(X) = \frac{2}{L} \left( \sin \left( \frac{n \pi}{L} X \right) \right)^2 . $$


Donde la parte exponencial, es como una fase global del número complejo, algo que no se ve reflejado en el observable físico, mostrando así una característica fundamental de los autoestados, y es que estos no evolucionan en el tiempo.

Si bien se ha encontrado la probabilidad de encontrar a la partícula en una posición específica, no se ha considerado el reservorio, ni el efecto de la temperatura que este ejerce sobre el sistema.

En física estadística se sabe que la densidad de probabilidad de tener a un sistema en un espacio determinado sometida a un reservorio con temperatura $T$ está íntimamente relacionada con la función partición $Z=\sum_{n=1}^{\infty} e^{-E_n / k_B T}$ que se explica desde en la teoría del marco canónico[falta cita]. Más aún el valor promedio de la densidad de probabilidad, denotado por $P_{th}(X,T)$, tendrá la forma

$$ P_{th}(X, T) = \frac{\sum_{n=1}^{\infty} P_n(X) e^{-\frac{E_n}{k_B T}}}{\sum_{n=1}^{\infty} e^{-E_n / k_B T}} = \frac{\sum_{n=1}^{\infty} \left[ \frac{2}{L} \left( \sin \left( \frac{n \pi}{L} X \right) \right)^2 \right] e^{-\frac{h^2 n^2}{8 m L^2 k_B T}}}{\sum_{n=1}^{\infty} e^{-\frac{h^2 n^2}{8 m L^2 k_B T}}} $$


Se pueden realizar los siguientes cambios de variables para un mejor tratamiento computacional


$$ T = t^* \frac{h^2}{8m L^2 k_B}, \qquad X = x^* \frac{L}{\pi}, \qquad \varphi_n(x) = \sqrt{\frac{2}{\pi}} \sin(n x) $$




Despejando estas nuevas temperatura $t$ y posición $x$ respectivamente, transforma
$P_{th}$ hacia la forma



$$ P_{th}(x, t) = \frac{2}{\pi} \frac{\sum_{n=1}^{\infty} e^{-\frac{n^2}{t}} (\sin(n x))^2}{\sum_{n=1}^{\infty} e^{-\frac{n^2}{t}}} . $$



Ahora, como no se puede computacionalmente evaluar dicha sumatoria hasta el infinito, se mirará el error al adicionar términos consecutivos, para ver en qué momento truncar la sumatoria, con un valor fijo de la temperatura y posición

```python
#@title Gráficos de convergencia

def Pth(x, t, n_max=100):
    numerator = 0.0
    denominator = 0.0

    for n in range(1, n_max + 1):
        exp_term = np.exp(-(n**2) / t)
        denominator += exp_term
        numerator += exp_term * (np.sin(n * x))**2

    constant = 2 / np.pi

    if denominator == 0:
        return 0

    return constant * (numerator / denominator)

x_fijo = np.pi / 2
t_values = [1.0, 50.0, 200.0, 500.0]
max_pasos = 50

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

for t in t_values:
    vals_n = []
    vals_pth = []
    vals_diff = []
    val_anterior = 0

    for n in range(1, max_pasos + 1):
        val_actual = Pth(x_fijo, t, n_max=n)
        vals_n.append(n)
        vals_pth.append(val_actual)

        if n == 1:
            vals_diff.append(0)
        else:
            vals_diff.append(abs(val_actual - val_anterior))

        val_anterior = val_actual

    ax1.plot(vals_n, vals_pth, marker='.', label=f't = {t}')
    ax2.plot(vals_n[1:], vals_diff[1:], marker='.', label=f't = {t}')

ax1.set_title('Convergencia de Pth')
ax1.set_xlabel('n')
ax1.set_ylabel('Valor de Pth')
ax1.grid(True)
ax1.legend()

ax2.set_title('Diferencia absoluta')
ax2.set_xlabel('n')
ax2.set_ylabel('|Pth(n) - Pth(n-1)|')
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

Se observa que a medida que aumentamos el valor de la temperatura, varía la necesidad de obtener un valor más grande de n. Sin embargo, se observa que para un valor de n=25, "es suficiente" para temperaturas grandes

Por otro lado tiene sentido pensar que en los bordes la densidad de probabilidad tienda a ser nula debido a las condiciones de frontera, sin embargo, ¿qué sucederá cuando se cambia la temperatura? ¿Cambiará la posición en el cual la densidad de probabilidad empiece a tender a cero?


Para esto se puede ver el comportamiento de la función de densidad de probabilidad en dos formas: la primera es observando cómo varía la probabilidad en función de la posición para diferentes temperaturas, la segunda es viendo cómo varía esta tendencia a la nulidad en función de la temperatura. Para este segundo caso, se considerará que la nulidad es cuando se alcanza el valor $P_{th}=0.1$, y se variará la temperatura, midiendo así la posición en la que se alcanza esta probabilidad.

```python
#@title Gráfico de densidad de probabilidad y ancho límite en función de la temperatura

def Pth(x, t, n_max=100):
    if isinstance(x, np.ndarray):
        numerator = np.zeros_like(x, dtype=float)
    else:
        numerator = 0.0

    denominator = 0.0

    for n in range(1, n_max + 1):
        exp_term = np.exp(-(n**2) / t)
        denominator += exp_term
        numerator += exp_term * (np.sin(n * x))**2

    constant = 2 / np.pi

    if denominator == 0:
        return 0

    return constant * (numerator / denominator)

x_values = np.linspace(0, np.pi, 500)

y_t1 = Pth(x_values, t=1.0)
y_t10 = Pth(x_values, t=10.0)
y_t100 = Pth(x_values, t=100.0)

t_range = np.linspace(1, 150, 100)
d_values = []

def target_func(x_guess, current_t):
    return Pth(x_guess, current_t) - 0.1

for t_val in t_range:
    try:
        d = brentq(target_func, 0.0001, 1.5, args=(t_val,))
        d_values.append(d)
    except ValueError:
        d_values.append(np.nan)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(x_values, y_t1, linestyle=':', color='black', label='(I) t = 1')
ax1.plot(x_values, y_t10, linestyle='--', color='black', label='(II) t = 10')
ax1.plot(x_values, y_t100, linestyle='-', color='black', label='(III) t = 100')

ax1.set_title('Densidad de probabilidad')
ax1.set_xlabel('Posición x')
ax1.set_ylabel('Pth(x, t)')
ax1.set_xlim(0, np.pi)
ax1.set_ylim(bottom=0)
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(t_range, d_values, color='black', linewidth=1.5)

ax2.set_title('Ancho d de la capa límite')
ax2.set_xlabel('Temperatura t')
ax2.set_ylabel('Ancho d (donde Pth = 0.1)')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Claramente se observa que tiene una forma simétrica, lo cual tiene mucho sentido si se piensa clásicamente. Por otro lado, se observa que al aumentar la temperatura se ve una forma más cuadrada de la función de densidad de probabilidad, esto debido a que entre mayor temperatura, la restricción de las condiciones de borde afectarán cada vez más, se puede pensar que la temperatura hace más uniforme la densidad de probabilidad.

## Dos partículas en una caja con potencial infinito (sin spin)


En la anterior situación la función de onda $\Psi$ adquirió una forma simple debido a las hipótesis y el modelo en sí mismo. Sin embargo, este modelo no es tan cercano a la realidad. Se le pueden agregar diferentes características a los sistemas, como poniéndole spin o cambiando el número de partículas. El siguiente escenario será uno en el que hay dos partículas sin spin.



Sin embargo, existen dos escenarios sobre los cuales pueden existir estas partículas, pueden ser fermiones o bosones. Para el caso de fermiones la función debe de ser antisimétrica bajo el intercambio de partículas, lo cual significa que no pueden ocupar el mismo nivel de energía es decir $n_{1}\neq n_{2}$, sea $\Psi^{f}_{n1,n2}$ la función de densidad de todo el sistema, claramente para que sea antisimétrica debe de cumplir que

$$ \Psi^f_{n1, n2}(x_1, x_2, t) = \frac{1}{\sqrt{2}} \left( \Psi_{n1}(x_1, t) \Psi_{n2}(x_2, t) - \Psi_{n1}(x_2, t) \Psi_{n2}(x_1, t) \right) $$


Sin embargo, hay que observar que solo contribuye la parte espacial de los fermiones, esto debido a lo siguiente


$$ \Psi(x_1, x_2, t)^{f} = \left[ \frac{1}{\sqrt{2}}(\varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1)) \right] e^{-i \frac{(E_{n1} + E_{n2})t}{\hbar}} $$



Claramente es una fase global que al momento de sacar la densidad de probabilidad, no cambiará nada en el resultado.




Por otro lado y de forma análoga a la anterior, se puede encontrar una expresión en el caso de la densidad de probabilidad denotada por $\Psi^{b}_{n1,n2}$. Solo hay que tener en cuenta que para estas partículas, si se admite la posibilidad de que ambas ocupen el mismo nivel energético. Así, la función de onda es simétrica y no importaría nuevamente la parte temporal. Por tanto, se tendría que adquiere la forma


$$ \psi^b_{n1, n2}(x_1, x_2) = \frac{1}{\sqrt{2}} \left( \varphi_{n1}(x_1) \varphi_{n2}(x_2) + \varphi_{n1}(x_2) \varphi_{n2}(x_1) \right). $$



De estas funciones de onda se deduce rápidamente las funciones de densidad de probabilidad. Para los fermiones se tendría

$$
\begin{aligned}
P^f_{n1, n2}(x_1, x_2) &= \frac{1}{2} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right]^2 \\
&= \frac{1}{2} \left[ \left(\sqrt{\frac{2}{\pi}} \sin(n_1 x_1) \cdot \sqrt{\frac{2}{\pi}} \sin(n_2 x_2)\right) - \left(\sqrt{\frac{2}{\pi}} \sin(n_1 x_2) \cdot \sqrt{\frac{2}{\pi}} \sin(n_2 x_1)\right) \right]^2 \\
&= \frac{1}{2} \left[ \frac{2}{\pi} \left( \sin(n_1 x_1)\sin(n_2 x_2) - \sin(n_1 x_2)\sin(n_2 x_1) \right) \right]^2 \\
&= \frac{1}{2} \cdot \frac{4}{\pi^2} \left( \sin(n_1 x_1)\sin(n_2 x_2) - \sin(n_1 x_2)\sin(n_2 x_1) \right)^2 \\
&= \frac{2}{\pi^2} \left( \sin(n_1 x_1)\sin(n_2 x_2) - \sin(n_1 x_2)\sin(n_2 x_1) \right)^2
\end{aligned}
$$



Por otro lado y de forma similar se puede encontrar una fórmula para los bosones


$$
\begin{aligned}
P^b_{n1, n2}(x_1, x_2) &= \frac{1}{2} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) + \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right]^2 \\
&= \frac{1}{2} \left[ \left(\sqrt{\frac{2}{\pi}} \sin(n_1 x_1) \cdot \sqrt{\frac{2}{\pi}} \sin(n_2 x_2)\right) + \left(\sqrt{\frac{2}{\pi}} \sin(n_1 x_2) \cdot \sqrt{\frac{2}{\pi}} \sin(n_2 x_1)\right) \right]^2 \\
&= \frac{1}{2} \left[ \frac{2}{\pi} \left( \sin(n_1 x_1)\sin(n_2 x_2) + \sin(n_1 x_2)\sin(n_2 x_1) \right) \right]^2 \\
&= \frac{1}{2} \cdot \frac{4}{\pi^2} \left( \sin(n_1 x_1)\sin(n_2 x_2) + \sin(n_1 x_2)\sin(n_2 x_1) \right)^2 \\
&= \frac{2}{\pi^2} \left( \sin(n_1 x_1)\sin(n_2 x_2) + \sin(n_1 x_2)\sin(n_2 x_1) \right)^2
\end{aligned}
$$

Así, se pueden escoger arbitrariamente valores de $n_{1}$ y $n_{2}$ y visualizar el comportamiento de las densidades de probabilidades.
Como los fermiones no pueden tener el mismo número, entonces se tomarán $n_{1}=n_{2}$ que es el estado base. Por otro lado, para los bosones se puede considerar $n=1$ y $n=5$, así se pueden visualizar en la siguiente forma


```python
#@title Código de probabilidades sin considerar temperatura

def Pb_bosones(x1, x2, n1, n2):
    prefactor = 2 / (np.pi**2)
    term_sym = np.sin(n1 * x1) * np.sin(n2 * x2) + np.sin(n1 * x2) * np.sin(n2 * x1)
    return prefactor * (term_sym**2)

L = np.pi
x = np.linspace(0, L, 100)
X1, X2 = np.meshgrid(x, x)

Z_a = Pb_bosones(X1, X2, n1=1, n2=1)
Z_b = Pb_bosones(X1, X2, n1=1, n2=5)

fig = plt.figure(figsize=(14, 12))

ax1 = fig.add_subplot(2, 2, 1, projection='3d')

surf1 = ax1.plot_surface(X1, X2, Z_a, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=5, cstride=5, alpha=0.9)

ax1.set_title('(a) Bosones: n1=1, n2=1')
ax1.set_xlabel('x1')
ax1.set_ylabel('x2')
ax1.set_zlabel('P')
ax1.set_xlim(0, 3.2)
ax1.set_ylim(0, 3.2)
ax1.set_zlim(0, 1.0)
ax1.view_init(elev=35, azim=45)

ax2 = fig.add_subplot(2, 2, 2, projection='3d')

surf2 = ax2.plot_surface(X1, X2, Z_b, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=3, cstride=3, alpha=0.9)

ax2.set_title('(b) Bosones: n1=1, n2=5')
ax2.set_xlabel('x1')
ax2.set_ylabel('x2')
ax2.set_zlabel('P')
ax2.set_xlim(0, 3.2)
ax2.set_ylim(0, 3.2)
ax2.set_zlim(0, 1.0)
ax2.view_init(elev=35, azim=45)

# Mapas de calor
ax3 = fig.add_subplot(2, 2, 3)
contour1 = ax3.contourf(X1, X2, Z_a, cmap='Wistia', levels=20)
fig.colorbar(contour1, ax=ax3)
ax3.set_title('(c) Mapa de calor: n1=1, n2=1')
ax3.set_xlabel('x1')
ax3.set_ylabel('x2')

ax4 = fig.add_subplot(2, 2, 4)
contour2 = ax4.contourf(X1, X2, Z_b, cmap='Wistia', levels=20)
fig.colorbar(contour2, ax=ax4)
ax4.set_title('(d) Mapa de calor: n1=1, n2=5')
ax4.set_xlabel('x1')
ax4.set_ylabel('x2')

fig.suptitle('Gráfico de bosones, sin temperatura y sin espín', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

Claramente el mapa de calor(o de nivel) meustra una forma alternativa de analizar la simetria de dichas densidad de probabilidad, pero se tendría que cuantificar las variaciones de una forma diferente. Si se considera prudente se hará un analisis mediante estos mapas de nivel.

Sin embargo, utilizando nuevamente las ideas de la física estadística, para encontrar una densidad de probabilidad en ambos sistemas, sabiendo que está expuesto a un reservorio, cada valor de probabilidad posible para sus estados n-ésimos se tendrán que multiplicar por una exponencial que contenga la información de ambos, es decir multiplicar por el factor de Boltzmann. Para normalizar dicha densidad de probabilidad se debe dividir por la función partición de dicho sistema. Por tanto, se tendría

$$ P_{th}^{f,b}(x_1, x_2, t) = \frac{\sum_{n_1=1}^{\infty} \sum_{n_2=1}^{\infty} P^{f,b}_{n_1, n_2}(x_1, x_2) \exp \left[ -\frac{E_{n_1, n_2}}{t} \right]}{\sum_{n_1=1}^{\infty} \sum_{n_2=1}^{\infty} \exp \left[ -\frac{E_{n_1, n_2}}{t} \right]}. $$



Donde $E_{n_{1},n_{2}}=n_{1}^{2}+n_{2}^{2}$

Hay que tener en consideración que computacionalmente se tienen que truncar nuevamente los valores de $n_{1},n_{2}$. Suponiendo que hay un valor para el que llegan a un máximo llamado $n_{max}$ entonces se pueden encontrar en ambos casos a partir de qué valores la contribución de nuevos términos afecta la densidad de probabilidad. Claramente como para los fermiones no es permitido valores de $n_{1},n_{2}$, entonces $1 \leq n_{1} \leq n_{max}-1$ y $n_{1}+1 \leq n_{2} \leq n_{max}$.

Así, veamos qué sucede con los valores que pueden tomar los $n's$

```python
#@title Código para encontrar los n_{máximos}

x1_fijo = np.pi / 3
x2_fijo = 2 * np.pi / 3
t_fijo = 100
max_n = 30

n_vals = np.arange(1, max_n + 1)
N1, N2 = np.meshgrid(n_vals, n_vals)

Z_fermion = np.zeros_like(N1, dtype=float)
Z_boson = np.zeros_like(N1, dtype=float)

for i in range(max_n):
    for j in range(max_n):
        n1 = N1[i, j]
        n2 = N2[i, j]

        exp_factor = np.exp(-(n1**2 + n2**2) / t_fijo)

        if n1 < n2:
            sin_f = np.sin(n1 * x1_fijo) * np.sin(n2 * x2_fijo) - np.sin(n1 * x2_fijo) * np.sin(n2 * x1_fijo)
            Z_fermion[i, j] = (sin_f**2) * exp_factor
        else:
            Z_fermion[i, j] = 0.0

        sin_b = np.sin(n1 * x1_fijo) * np.sin(n2 * x2_fijo) + np.sin(n1 * x2_fijo) * np.sin(n2 * x1_fijo)
        Z_boson[i, j] = (sin_b**2) * exp_factor

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(N1, N2, Z_fermion, cmap='Blues', edgecolor='k', linewidth=0.2, alpha=0.8)

ax1.set_title(f'Error/Contribución FERMIONES (t={t_fijo})')
ax1.set_xlabel('$n_1$')
ax1.set_ylabel('$n_2$')
ax1.set_zlabel('Magnitud del Error')
ax1.view_init(elev=30, azim=-45)

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(N1, N2, Z_boson, cmap='Oranges', edgecolor='k', linewidth=0.2, alpha=0.8)
ax2.set_title(f'Error/Contribución BOSONES (t={t_fijo})')
ax2.set_xlabel('$n_1$')
ax2.set_ylabel('$n_2$')
ax2.set_zlabel('Magnitud del Error')
ax2.view_init(elev=30, azim=-45)

fig.suptitle('Error/contribución', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

Claramente para valores mayores a los n's iguales a 15 es suficiente en ambos casos para obtener un error menor a 0.1, por tanto se tomará el valor tomado para el caso de 1 partícula, es decir $n_{max}=25$, asegurándonos que se tendrán suficientes términos de esta forma. ## Partículas con spin

Por último, se puede analizar el caso de partículas con spin. Para esta situación se considerará fermiones que tienen spin 1/2. Considerando que la forma completa de la función de onda es

$$ \Psi_{total} = \psi_{espacial} \times \chi_{spin} \times \phi_{temporal} $$

Teniendo en cuenta que adicionalmente se tendría que cumplir que debe de ser una función antisimétrica, entonces todas las posibles combinaciones que admite será, una parte espacial simétrica y una de spin antisimétrica, o viceversa(la temporal se volvería a anular como en el caso de dos partículas sin spin). Así, admitirá en el primer caso

$$ \Psi^f_{n1,n2} = \frac{1}{\sqrt{2}} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) + \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right] \frac{1}{\sqrt{2}} \left[ \chi_+(1)\chi_-(2) - \chi_+(2)\chi_-(1) \right], $$


Por otro lado, en el caso de que la parte espacial antisimétrica, existen tres posibilidades de combinar el spin, estas formas serían



$$
\begin{aligned}
\Psi^f_{n1,n2} &= \frac{1}{\sqrt{2}} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right] \frac{1}{\sqrt{2}} \left[ \chi_+(1)\chi_-(2) + \chi_+(2)\chi_-(1) \right], \\[10pt]
\Psi^f_{n1,n2} &= \frac{1}{\sqrt{2}} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right] [\chi_+(1)\chi_+(2)], \\[10pt]
\Psi^f_{n1,n2} &= \frac{1}{\sqrt{2}} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right] [\chi_-(1)\chi_-(2)].
\end{aligned}
$$


Para poder encontrar, la densidad de probabilidad asociada, hay que considerar la cantidad de casos que hay para cada situación, y se realiza un promedio ponderado sumando la contribución del caso simétrico espacial(que tiene peso 1/4) y la contribución del caso antisimétrico espacial (de peso 3/4) obteniendo así que la probabilidad tendrá la forma



$$
\begin{aligned}
P_{n1,n2}(x_1, x_2) &= \frac{1}{4} \left( \frac{1}{2} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) + \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right]^2 \right) \\
&\quad + \frac{3}{4} \left( \frac{1}{2} \left[ \varphi_{n1}(x_1)\varphi_{n2}(x_2) - \varphi_{n1}(x_2)\varphi_{n2}(x_1) \right]^2 \right).
\end{aligned}
$$




Por otro lado se podría pensar en la posibilidad de admitir los bosones con spin, sin embargo, al ser simétrica la función la función propia de ella, no tiene sentido pensar en incluirle el spin para que la función total de onda tenga una forma simétrica. De hecho, existe otra combinación, y es pensar que ambas partes, la espacial y la asociada al spin tiene forma antisimétrica, sin embargo, para este caso como se admite la posibilidad de tener energía con estados excitados iguales, la función asociada a la posición tendría problemas, esto debido a la forma sinusoidal. Más aún, como el sistema de bosones tiende a ocupar el mismo nivel de energía debido a la condensación de Bose-Einstein no tiene sentido considerar una función de onda asociada antisimétrica.

Por otro lado, como los fermiones sí admiten la probabilidad anteriormente expuesta, también admitirá de forma análoga una densidad de probabilidad en una situación en la que se está sometida a un reservorio con temperatura $T$, se tendría que multiplicar por el factor de Boltzmann además de sumar sobre todas las posibilidades y dividir sobre la función partición. Así se tendría la siguiente ecuación


$$ P_{th}(x_1, x_2, t) = \frac{\sum_{n_1=1}^{n_{\text{max}}} \sum_{n_2=n_1}^{n_{\text{max}}} P_{n_1, n_2}(x_1, x_2) e^{-\frac{n_1^2+n_2^2}{t}}}{\sum_{n_1=1}^{n_{\text{max}}} \sum_{n_2=n_1}^{n_{\text{max}}} e^{-\frac{n_1^2+n_2^2}{t}}} $$

```python
#@title Código para mostrar la densidad de probabilidad de fermiones con reservorio y sin reservorio

L = np.pi
points = 80
x = np.linspace(0, L, points)
X1, X2 = np.meshgrid(x, x)
const_wave = 2 / (np.pi**2)

def get_P_spin_fermion(n1, n2, X1, X2):
    p1x1 = np.sin(n1 * X1)
    p2x2 = np.sin(n2 * X2)
    p1x2 = np.sin(n1 * X2)
    p2x1 = np.sin(n2 * X1)

    term_sym = 0.5 * (p1x1 * p2x2 + p1x2 * p2x1)**2
    term_anti = 0.5 * (p1x1 * p2x2 - p1x2 * p2x1)**2

    P_total = const_wave * (0.25 * term_sym + 0.75 * term_anti)

    return P_total

Z_pure = get_P_spin_fermion(1, 2, X1, X2)

t = 10.0
n_max = 30
num_sum = np.zeros_like(X1)
den_sum = 0.0

for n1 in range(1, n_max + 1):
    for n2 in range(n1, n_max + 1):

        E = n1**2 + n2**2
        boltz = np.exp(-E / t)

        P_state = get_P_spin_fermion(n1, n2, X1, X2)

        num_sum += P_state * boltz
        den_sum += boltz

Z_thermal = num_sum / den_sum

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X1, X2, Z_pure, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=4, cstride=4, alpha=0.9)

ax1.set_title('Sin Reservorio (Estado n1=1, n2=2)\nFermiones con Espín')
ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.set_zlabel('$P$')
ax1.set_xlim(0, 3.2)
ax1.set_ylim(0, 3.2)
ax1.view_init(elev=35, azim=45)

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X1, X2, Z_thermal, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=4, cstride=4, alpha=0.9)

ax2.set_title('Con Reservorio (t=10)\nFermiones con Espín')
ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.set_zlabel('$P_{th}$')
ax2.set_xlim(0, 3.2)
ax2.set_ylim(0, 3.2)
ax2.view_init(elev=35, azim=45)

fig.suptitle('Comparativa de Fermiones con espín al ser sometido a un reservorio', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

# Pozo de potencial finito


Para esta situación la ecuación que se debe de resolver de evolución temporal de Schrödinger cambia, de hecho cambian las condiciones de frontera, algo que influye directamente en las soluciones que se encuentran en dicho problema. Considerando un pozo de ancho L, es decir que va desde -L/2 hasta L/2 tendrá un potencial de la siguiente forma

$$
V(x) = \begin{cases}
   0 & \text{si } -L/2 < x < L/2 \quad (\text{Región I: Dentro}) \\
   V_0 & \text{si } |x| \geq L/2 \quad (\text{Región II: Fuera})
\end{cases}
$$




Para esta situación, se buscan estos ligados es decir, estados donde la energía cumpla que $E<V_{0}$. La ecuación general de Schrödinger admite la siguiente reescritura


$$
-\frac{\hbar^2}{2m} \frac{d^2\psi}{dx^2} + V(x)\psi = E\psi
$$

Sin embargo, como la solución debe de ser consistente tanto dentro el pozo como fuera de este, entonces el problema debe de dividirse en dos partes.


Partícula dentro del pozo:
Aquí $V(x) = 0$

La ecuación es:

$$
\frac{d^2\psi}{dx^2} = -\frac{2mE}{\hbar^2}\psi = -k^2\psi
$$

Definimos el número de onda $k$ (que está relacionado con el momento lineal $p=\hbar k$):

$$
k = \frac{\sqrt{2mE}}{\hbar}
$$

Las soluciones matemáticas son ondas oscilatorias: senos y cosenos.



Partícula Fuera del pozo:

Como fuera del pozo se debe de cumplir que $V(x)=V_{0}$, el término cinético es negativo(aunque esto de forma clásica no se puede cumplir, cuánticamente sí está permitido)


$$
\frac{d^2\psi}{dx^2} = \frac{2m(V_0 - E)}{\hbar^2}\psi = \kappa^2\psi
$$

Definimos la constante de decaimiento $\kappa$:

$$
\kappa = \frac{\sqrt{2m(V_0 - E)}}{\hbar}
$$

Las soluciones son exponenciales reales: $e^{\kappa x}$ y $e^{-\kappa x}$.

*   A la derecha ($x > L/2$), la función no puede crecer al infinito, así que solo nos sirve $e^{-\kappa x}$.
*   A la izquierda ($x < -L/2$), solo nos sirve $e^{\kappa x}$.


Como el pozo es simétrico alrededor del cero, las soluciones naturales prefieren ser pares. Para los estados pares $\psi(x)=\psi(-x)$ debe de tener una solución de cosenos y claramente para los impares $\psi(x)=-\psi(-x)$ se usan senos. Así, la solución debería de cumplir con esto


$$
\psi(x) = \begin{cases}
   A \cos(kx) & \text{dentro } (|x| < L/2) \\
   B e^{-\kappa x} & \text{fuera (lado derecho } x > L/2)
\end{cases}
$$


Por la continuidad que debe de cumplir se cumplirá

$$
A \cos(kL/2) = B e^{-\kappa L/2}
$$

Además de la continuidad de la pendiente en $x=L/2$ se cumplirá que


$$
-kA \sin(kL/2) = -\kappa B e^{-\kappa L/2}
$$

Combinando ambas se tendría la ecuación fundamental


$$
\frac{-kA \sin(kL/2)}{A \cos(kL/2)} = \frac{-\kappa B e^{-\kappa L/2}}{B e^{-\kappa L/2}}
$$


Por lo tanto

$$
k \tan \left( \frac{kL}{2} \right) = \kappa
$$


Que al reemplazar las definiciones de las constantes queda de la forma

$$
\sqrt{\frac{2mE}{\hbar^2}} \tan \left( \sqrt{\frac{2mE}{\hbar^2}} \frac{L}{2} \right) = \sqrt{\frac{2m(V_0 - E)}{\hbar^2}}
$$



Se puede simplificar de la siguiente forma

$$
\tan(z) = \sqrt{\left(\frac{z_0}{z}\right)^2 - 1}
$$

Donde las definiciones de $z$ y $z_0$ son:

*   $$ z = \frac{kL}{2} $$
*   $$ z_0 = \frac{L}{2\hbar}\sqrt{2mV_0} $$



Claramente la variable $z$ es la que tiene la información de la energía, para ser más preciso es proporcional a $\sqrt{E}$

Para los estados impares $\psi(x)=-\psi(-x)$ se usan senos dentro del pozo. Así, la solución cumplirá con esto:

$$
\psi(x) = \begin{cases}
   A \sin(kx) & \text{dentro } (|x| < L/2) \\
   B e^{-\kappa x} & \text{fuera (lado derecho } x > L/2)
\end{cases}
$$

Por la continuidad que debe de cumplir la función en $x=L/2$ se cumplirá:

$$
A \sin(kL/2) = B e^{-\kappa L/2}
$$

Además de la continuidad de la pendiente en $x=L/2$ se cumplirá que:

$$
kA \cos(kL/2) = -\kappa B e^{-\kappa L/2}
$$

Combinando ambas se tendría la ecuación fundamental:

$$
\frac{kA \cos(kL/2)}{A \sin(kL/2)} = \frac{-\kappa B e^{-\kappa L/2}}{B e^{-\kappa L/2}}
$$

Por lo tanto:

$$
k \cot \left( \frac{kL}{2} \right) = -\kappa
$$

O reordenando para dejar el término $\kappa$ positivo:

$$
-k \cot \left( \frac{kL}{2} \right) = \kappa
$$

Que al reemplazar las definiciones de las constantes queda de la forma:

$$
-\sqrt{\frac{2mE}{\hbar^2}} \cot \left( \sqrt{\frac{2mE}{\hbar^2}} \frac{L}{2} \right) = \sqrt{\frac{2m(V_0 - E)}{\hbar^2}}
$$

Se puede simplificar de la siguiente forma:

$$
-\cot(z) = \sqrt{\left(\frac{z_0}{z}\right)^2 - 1}
$$






Así, claramente lo que se necesita es encontrar las energías que $E_{n1},E_{n2}...$ la cantidad de energías que se permiten es hasta el n máximo que permite el potencial, y la estructura del pozo más precisamente el límite que da la fracción $z_{0}/z$. Estas raíces de estas ecuaciones se encuentran y se utilizan para describir a las funciones de onda. De forma análoga al potencial infinito la parte temporal no actúa sobre el sistema, y se considerará un sistema en el cual no hay spin.

```python
#@title Código para encontrar las energías de una partícula en un pozo finito

def encontrar_energias_pozo_finito():
    hbar = 1.0
    m = 1.0
    L = 1.0
    V0 = 400.0

    z0 = (L / (2 * hbar)) * np.sqrt(2 * m * V0)

    energias = []
    n = 1

    print(f"--- Configuración del Pozo ---")
    print(f"Ancho L = {L}")
    print(f"Altura V0 = {V0}")
    print(f"Parámetro z0 = {z0:.4f}")
    print("-" * 30)
    print("ENERGÍAS ENCONTRADAS:")

    while True:
        z_min = (n - 1) * np.pi / 2
        z_max = n * np.pi / 2

        if z_min >= z0:
            break

        epsilon = 1e-5
        limite_inferior = z_min + epsilon
        limite_superior = min(z_max, z0) - epsilon

        if limite_inferior >= limite_superior:
            break

        root_z = None

        try:
            if n % 2 != 0:
                def funcion_objetivo(z):
                    return np.tan(z) - np.sqrt((z0/z)**2 - 1)
            else:
                def funcion_objetivo(z):
                    return -1/np.tan(z) - np.sqrt((z0/z)**2 - 1)

            root_z = brentq(funcion_objetivo, limite_inferior, limite_superior)

            E = (2 * hbar**2 * root_z**2) / (m * L**2)

            energias.append(E)
            print(f"E_{{{n}}} = {E:.6f}")

        except ValueError:
            pass

        n += 1

    return energias

energias_encontradas = encontrar_energias_pozo_finito()
```

Por último, para encontrar una expresión completa para las funciones de probabilidad, es requerido las constantes A y B que aparecen en las funciones. Claramente dependerá si se está trabajando con un n par o impar.
Hay que notar que se llega rápidamente de las condiciones de frontera que

$$
B = A \left( \frac{\cos(ka)}{e^{-\kappa a}} \right) = A \cos(ka)e^{\kappa a}
$$

Por tanto, una vez fijada una, se encontraría la otra. Sin embargo, se requiere otra condición para encontrar una de las dos variables. La ecuación que nos proporciona qué valor puede tomar una de estas constantes es de hecho la condición de normalización que deben de cumplir dichas funciones de onda. Por hipótesis, en mecánica cuántica se debe de cumplir que




$$
\int_{-\infty}^{\infty} |\psi(x)|^2 dx = 1
$$


Como la integral es de hecho simétrica, se puede dividir en zonas, permitiendo así una expresión para $\psi$

$$
\underbrace{\int_{-a}^{a} |A \cos(kx)|^2 dx}_{\text{Dentro}} + \underbrace{2 \int_{a}^{\infty} |B e^{-\kappa x}|^2 dx}_{\text{Fuera}} = 1
$$

Por tanto, quedaría de la siguiente forma al despejar

$$
A^2 \left[ \int_{-a}^{a} \cos^2(kx)dx + 2 \cos^2(ka)e^{2\kappa a} \int_{a}^{\infty} e^{-2\kappa x}dx \right] = 1
$$


$$
A = \frac{1}{\sqrt{\text{I}}}
$$

Donde I es la integral numérica, que tiene como implícita los valores del n a tratar como información dentro de $k,κ$. Así quedarían descritos completamente los valores necesarios para encontrar a dichas funciones

```python
#@title Función para las funciones de onda normalizadas

def generar_funciones_onda(x_grid, energias, L, V0, m=1.0, hbar=1.0):
    lista_psi = []
    a = L / 2

    for i, E in enumerate(energias):
        n = i + 1

        k = np.sqrt(2 * m * E) / hbar

        kappa = np.sqrt(2 * m * (V0 - E)) / hbar

        psi = np.zeros_like(x_grid)

        mask_dentro = np.abs(x_grid) <= a
        mask_derecha = x_grid > a
        mask_izquierda = x_grid < -a

        if n % 2 != 0:
            psi[mask_dentro] = np.cos(k * x_grid[mask_dentro])

            val_borde = np.cos(k * a)

            psi[mask_derecha] = val_borde * np.exp(-kappa * (x_grid[mask_derecha] - a))
            psi[mask_izquierda] = val_borde * np.exp(-kappa * (-x_grid[mask_izquierda] - a))

        else:
            psi[mask_dentro] = np.sin(k * x_grid[mask_dentro])

            val_borde = np.sin(k * a)

            psi[mask_derecha] = val_borde * np.exp(-kappa * (x_grid[mask_derecha] - a))
            psi[mask_izquierda] = -val_borde * np.exp(-kappa * (-x_grid[mask_izquierda] - a))

        area_cuadrada = np.trapz(psi**2, x_grid)

        norm = 1.0 / np.sqrt(area_cuadrada)

        psi_normalizada = psi * norm

        lista_psi.append(psi_normalizada)

    return lista_psi

energias_ejemplo = [9.39, 37.45, 83.83, 148.06]
L_pozo = 1.0
V0_pozo = 400.0

x = np.linspace(-1.5, 1.5, 1000)

psis = generar_funciones_onda(x, energias_ejemplo, L_pozo, V0_pozo)

plt.figure(figsize=(10, 6))

for i in range(min(3, len(psis))):
    plt.plot(x, psis[i] * 5 + energias_ejemplo[i], label=f'n={i+1} (E={energias_ejemplo[i]:.2f})', linewidth=2)

    plt.hlines(energias_ejemplo[i], -1.5, 1.5, linestyle=':', color='gray', alpha=0.5)

plt.vlines([-0.5, 0.5], 0, 100, color='black', linewidth=3, label='Paredes Pozo')

plt.title('Funciones de Onda Normalizadas (Pozo Finito)')
plt.xlabel('Posición x')
plt.ylabel('$\Psi_n(x)$ (Desplazada por Energía)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("--- Verificación de Normalización ---")
for i, psi in enumerate(psis):
    integ = np.trapz(psi**2, x)
    print(f"Estado n={i+1}: Integral |psi|^2 = {integ:.5f}")
```

Para poder encontrar las densidades de probabilidades, se hace de nuevo la sumatoria con los factores de Boltzmann. Claramente hay que tener en cuenta qué valores $n_{max}$ es permitido, esto dependiendo de la altura del pozo, así la función de densidad de probabilidad estaría dada por

$$
P_{th}(x, t) = \frac{\sum_{n=1}^{N_{max}} |\psi_n(x)|^2 e^{-E_n/t}}{\sum_{n=1}^{N_{max}} e^{-E_n/t}}
$$

Dependiendo de si n es par o impar se usará una función de onda asociada. Es posible por supuesto que a mayor temperatura se encuentre la necesidad de dar más valores de $n's$ de la serie, sin embargo para valores muy altos de temperatura es posible que no haya necesidad de extender a muchos valores de n, pues como se sabe, existe la posibilidad de que la sumatoria no aporte más a la sumatoria

```python
#@title Código para la densidad de probabilidad con la termalización y convergencia al aumentar términos variando la temperatura

def encontrar_energias(L, V0, m=1.0, hbar=1.0):
    z0 = (L / (2 * hbar)) * np.sqrt(2 * m * V0)
    energias = []
    n = 1
    while True:
        z_min, z_max = (n - 1) * np.pi / 2, n * np.pi / 2
        if z_min >= z0: break
        epsilon = 1e-5
        lim_inf, lim_sup = z_min + epsilon, min(z_max, z0) - epsilon
        if lim_inf >= lim_sup: break
        try:
            if n % 2 != 0:
                f = lambda z: np.tan(z) - np.sqrt((z0/z)**2 - 1)
            else:
                f = lambda z: -1/np.tan(z) - np.sqrt((z0/z)**2 - 1)
            root = brentq(f, lim_inf, lim_sup)
            energias.append((2 * hbar**2 * root**2) / (m * L**2))
        except ValueError: pass
        n += 1
    return energias

def generar_psis(x_grid, energias, L, V0, m=1.0, hbar=1.0):
    psis = []
    a = L / 2
    for i, E in enumerate(energias):
        n = i + 1
        k = np.sqrt(2 * m * E) / hbar
        kappa = np.sqrt(2 * m * (V0 - E)) / hbar
        psi = np.zeros_like(x_grid)
        mask_in = np.abs(x_grid) <= a
        mask_r = x_grid > a
        mask_l = x_grid < -a

        if n % 2 != 0:
            psi[mask_in] = np.cos(k * x_grid[mask_in])
            val = np.cos(k * a)
            psi[mask_r] = val * np.exp(-kappa * (x_grid[mask_r] - a))
            psi[mask_l] = val * np.exp(-kappa * (-x_grid[mask_l] - a))
        else:
            psi[mask_in] = np.sin(k * x_grid[mask_in])
            val = np.sin(k * a)
            psi[mask_r] = val * np.exp(-kappa * (x_grid[mask_r] - a))
            psi[mask_l] = -val * np.exp(-kappa * (-x_grid[mask_l] - a))

        area = np.trapz(psi**2, x_grid)
        norm = 1.0 / np.sqrt(area) if area > 0 else 0.0
        psis.append(psi * norm)
    return psis

L = 1.0
V0_visual = 400.0
T_visual = [5.0, 50.0, 200.0]
x_plot = np.linspace(-1.5, 1.5, 500)

energias_izq = encontrar_energias(L, V0_visual)
psis_izq = generar_psis(x_plot, energias_izq, L, V0_visual)

V0_conv = 10000.0
T_conv = [50.0, 500.0, 2000.0]

energias_der = encontrar_energias(L, V0_conv)
x_plot_conv = np.linspace(-1.5, 1.5, 2000)
psis_der = generar_psis(x_plot_conv, energias_der, L, V0_conv)
x_test = 0.1
idx_test = (np.abs(x_plot_conv - x_test)).argmin()
vals_psi_der = [p[idx_test] for p in psis_der]
num_estados_der = len(energias_der)

def calcular_Pth_final(psis_lista, energias_lista, T):
    num = np.zeros_like(psis_lista[0])
    den = 0.0
    for i, psi in enumerate(psis_lista):
        E = energias_lista[i]
        boltz = np.exp(-E/T)
        num += (psi**2) * boltz
        den += boltz
    return num/den if den > 0 else np.zeros_like(num)

def calcular_Pth_suma_parcial(vals_psi, energias, T, n):
    num, den = 0.0, 0.0
    for i in range(n):
        E = energias[i]
        v = vals_psi[i]
        boltz = np.exp(-E/T)
        num += (v**2)*boltz
        den += boltz
    return num/den if den > 0 else 0.0

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

estilos = [':', '--', '-']
for i, t in enumerate(T_visual):
    Pth = calcular_Pth_final(psis_izq, energias_izq, t)
    ax1.plot(x_plot, Pth, linestyle=estilos[i], linewidth=2, label=f'T = {t}')

ax1.axvline(-L/2, color='black', linewidth=2, label='Paredes')
ax1.axvline(L/2, color='black', linewidth=2)
ax1.set_title(f'Densidad de Probabilidad Térmica (Pozo Finito $V_0={V0_visual}$)')
ax1.set_xlabel('Posición x')
ax1.set_ylabel('$P_{th}(x, T)$')
ax1.set_xlim(-1.5, 1.5)
ax1.legend()
ax1.grid(True, alpha=0.3)

epsilon = 1e-20
for t in T_conv:
    diffs = []
    n_vals = []
    val_anterior = 0.0

    for n in range(1, num_estados_der + 1):
        val_actual = calcular_Pth_suma_parcial(vals_psi_der, energias_der, t, n)

        if n > 1:
            diferencia = abs(val_actual - val_anterior)
            if diferencia == 0: diferencia = epsilon
            diffs.append(diferencia)
            n_vals.append(n)

        val_anterior = val_actual

    ax2.plot(n_vals, diffs, '.-', label=f'T={t}', markersize=3, linewidth=1)

ax2.set_title('Convergencia extendida (Diferencia Absoluta)')
ax2.set_xlabel('Estados sumados (N)')
ax2.set_ylabel('|P(N) - P(N-1)|')
ax2.set_yscale('log')
ax2.legend()
ax2.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
```

Es decir, se debe de agregar el factor de Boltzmann de termalización y dividir sobre la función de partición.

```python
#@title Código para dos partículas en un potencial finito con termalización
import numpy as np
from scipy.optimize import brentq
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def encontrar_energias(L, V0, m=1.0, hbar=1.0):
    z0 = (L / (2 * hbar)) * np.sqrt(2 * m * V0)
    energias = []
    n = 1
    while True:
        z_min, z_max = (n - 1) * np.pi / 2, n * np.pi / 2
        if z_min >= z0: break
        epsilon = 1e-5
        lim_inf, lim_sup = z_min + epsilon, min(z_max, z0) - epsilon
        if lim_inf >= lim_sup: break
        try:
            if n % 2 != 0:
                f = lambda z: np.tan(z) - np.sqrt((z0/z)**2 - 1)
            else:
                f = lambda z: -1/np.tan(z) - np.sqrt((z0/z)**2 - 1)
            root = brentq(f, lim_inf, lim_sup)
            energias.append((2 * hbar**2 * root**2) / (m * L**2))
        except ValueError: pass
        n += 1
    return energias

def generar_psis(x_grid, energias, L, V0, m=1.0, hbar=1.0):
    psis = []
    a = L / 2
    for i, E in enumerate(energias):
        n = i + 1
        k = np.sqrt(2 * m * E) / hbar
        kappa = np.sqrt(2 * m * (V0 - E)) / hbar
        psi = np.zeros_like(x_grid)
        mask_in = np.abs(x_grid) <= a
        mask_r = x_grid > a
        mask_l = x_grid < -a
        if n % 2 != 0:
            psi[mask_in] = np.cos(k * x_grid[mask_in])
            val = np.cos(k * a)
            psi[mask_r] = val * np.exp(-kappa * (x_grid[mask_r] - a))
            psi[mask_l] = val * np.exp(-kappa * (-x_grid[mask_l] - a))
        else:
            psi[mask_in] = np.sin(k * x_grid[mask_in])
            val = np.sin(k * a)
            psi[mask_r] = val * np.exp(-kappa * (x_grid[mask_r] - a))
            psi[mask_l] = -val * np.exp(-kappa * (-x_grid[mask_l] - a))
        area = np.trapz(psi**2, x_grid)
        psis.append(psi / np.sqrt(area))
    return psis

def dibujar_contornos(ax, Z_data, L, x_vec):
    idx_pos = (np.abs(x_vec - L/2)).argmin()
    idx_neg = (np.abs(x_vec + L/2)).argmin()
    val_p, val_n = x_vec[idx_pos], x_vec[idx_neg]
    style = {'color': 'black', 'linewidth': 5, 'linestyle': ':', 'zorder': 10}
    ax.plot(np.full_like(x_vec, val_p), x_vec, Z_data[:, idx_pos], **style)
    ax.plot(np.full_like(x_vec, val_n), x_vec, Z_data[:, idx_neg], **style)
    ax.plot(x_vec, np.full_like(x_vec, val_p), Z_data[idx_pos, :], **style)
    ax.plot(x_vec, np.full_like(x_vec, val_n), Z_data[idx_neg, :], **style)
    k = L/2
    ax.plot([-k, -k, k, k, -k], [-k, k, k, -k, -k], [0,0,0,0,0], 'k:', lw=3, zorder=2)

L = 1.0
V0 = 50.0
T_sys = 50.0
x_vals = np.linspace(-1.0, 1.0, 100)
X1, X2 = np.meshgrid(x_vals, x_vals)
energias = encontrar_energias(L, V0)
psis_1d = generar_psis(x_vals, energias, L, V0)
num_estados = len(energias)
print(f"Se encontraron {num_estados} estados ligados en el pozo finito.")
Pth_boson = np.zeros_like(X1)
Z_boson = 0.0
Pth_fermion = np.zeros_like(X1)
Z_fermion = 0.0
for i in range(num_estados):
    for j in range(num_estados):
        E_total = energias[i] + energias[j]
        boltz = np.exp(-E_total / T_sys)
        psi_n1 = psis_1d[i]
        psi_n2 = psis_1d[j]
        prod_12 = np.outer(psi_n2, psi_n1)
        prod_21 = np.outer(psi_n1, psi_n2)
        if i <= j:
            if i == j:
                Psi_B = prod_12
                weight = 1.0
            else:
                Psi_B = (1/np.sqrt(2)) * (prod_12 + prod_21)
                weight = 1.0
            Prob_B = Psi_B**2
            Pth_boson += Prob_B * boltz
            Z_boson += boltz
        if i < j:
            Psi_F = (1/np.sqrt(2)) * (prod_12 - prod_21)
            Prob_F = Psi_F**2
            Pth_fermion += Prob_F * boltz
            Z_fermion += boltz
if Z_boson > 0:
    Pth_boson /= Z_boson
if Z_fermion > 0:
    Pth_fermion /= Z_fermion
fig = plt.figure(figsize=(14, 6))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X1, X2, Pth_boson, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, alpha=0.7, rstride=3, cstride=3, zorder=1)
dibujar_contornos(ax1, Pth_boson, L, x_vals)
ax1.set_title(f'Bosones en Pozo Finito ($V_0={V0}, T={T_sys}$)')
ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.set_zlabel('$P_{th}$')
ax1.view_init(elev=45, azim=45)
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X1, X2, Pth_fermion, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, alpha=0.7, rstride=3, cstride=3, zorder=1)
dibujar_contornos(ax2, Pth_fermion, L, x_vals)
ax2.set_title(f'Fermiones en Pozo Finito ($V_0={V0}, T={T_sys}$)')
ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.set_zlabel('$P_{th}$')
ax2.view_init(elev=45, azim=45)
fig.suptitle('Bosones y Fermiones sometidos a una temperatura (sin espín) (potencial finito)', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

```python
#@title Código para ver los bordes en las gráficas

L = 1.0
V0 = 40.0
T_sys = 5.0

puntos = 150
limit_vis = 1.2
x_vals = np.linspace(-limit_vis, limit_vis, puntos)
X1, X2 = np.meshgrid(x_vals, x_vals)

energias = encontrar_energias(L, V0)
psis_1d = generar_psis(x_vals, energias, L, V0)
num_estados = len(energias)

Pth_boson = np.zeros_like(X1)
Pth_fermion = np.zeros_like(X1)
Z_b, Z_f = 0.0, 0.0

for i in range(num_estados):
    for j in range(num_estados):
        E_tot = energias[i] + energias[j]
        boltz = np.exp(-E_tot / T_sys)

        p1 = psis_1d[i]
        p2 = psis_1d[j]

        psi_n1_grid = p1.reshape(1, -1)
        psi_n2_grid = p2.reshape(-1, 1)

        prod_12 = psi_n1_grid * psi_n2_grid

        psi_n2_grid_inv = p2.reshape(1, -1)
        psi_n1_grid_inv = p1.reshape(-1, 1)
        prod_21 = psi_n2_grid_inv * psi_n1_grid_inv

        if i <= j:
            Psi_B = prod_12 if i == j else (prod_12 + prod_21)/np.sqrt(2)
            Pth_boson += (Psi_B**2) * boltz
            Z_b += boltz

        if i < j:
            Psi_F = (prod_12 - prod_21)/np.sqrt(2)
            Pth_fermion += (Psi_F**2) * boltz
            Z_f += boltz

if Z_b > 0: Pth_boson /= Z_b
if Z_f > 0: Pth_fermion /= Z_f

def dibujar_contornos_paredes(ax, X1, X2, Z_data, L, x_vec):
    idx_L_pos = (np.abs(x_vec - L/2)).argmin()
    idx_L_neg = (np.abs(x_vec + L/2)).argmin()

    val_pos = x_vec[idx_L_pos]
    val_neg = x_vec[idx_L_neg]

    style_surf = {'color': 'black', 'linewidth': 3, 'linestyle': ':', 'zorder': 10}

    ax.plot(np.full_like(x_vec, val_pos), x_vec, Z_data[:, idx_L_pos], **style_surf, label='Borde del pozo')

    ax.plot(np.full_like(x_vec, val_neg), x_vec, Z_data[:, idx_L_neg], **style_surf)

    ax.plot(x_vec, np.full_like(x_vec, val_pos), Z_data[idx_L_pos, :], **style_surf)

    ax.plot(x_vec, np.full_like(x_vec, val_neg), Z_data[idx_L_neg, :], **style_surf)

    k = L/2
    ax.plot([-k, -k, k, k, -k], [-k, k, k, -k, -k], [0,0,0,0,0],
            color='black', linestyle=':', linewidth=2, zorder=2)

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X1, X2, Pth_boson, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, alpha=0.7, zorder=1, rstride=3, cstride=3)

dibujar_contornos_paredes(ax1, X1, X2, Pth_boson, L, x_vals)

ax1.set_title('Bosones')
ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.set_xlim(-limit_vis, limit_vis)
ax1.set_ylim(-limit_vis, limit_vis)
ax1.view_init(elev=35, azim=45)
ax1.legend()

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X1, X2, Pth_fermion, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, alpha=0.7, zorder=1, rstride=3, cstride=3)

dibujar_contornos_paredes(ax2, X1, X2, Pth_fermion, L, x_vals)

ax2.set_title('Fermiones')
ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.set_xlim(-limit_vis, limit_vis)
ax2.set_ylim(-limit_vis, limit_vis)
ax2.view_init(elev=35, azim=45)
ax2.legend()

fig.suptitle('Bosones y Fermiones sometidos a una temperatura (sin espín) (potencial finito)', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

Por último se pueden considerar el spin, y ver cómo influye este dentro de las formas de las probabilidades.

```python
def get_P_spin_finite(n1, n2, psis_1d):
    p1, p2 = psis_1d[n1 - 1], psis_1d[n2 - 1]

    psi_a_grid = p1.reshape(1, -1)
    psi_b_grid = p2.reshape(-1, 1)
    term1 = psi_a_grid * psi_b_grid

    psi_b_grid_inv = p2.reshape(1, -1)
    psi_a_grid_inv = p1.reshape(-1, 1)
    term2 = psi_a_grid_inv * psi_b_grid_inv

    Psi_sym = (term1 + term2) / np.sqrt(2)
    prob_sym = Psi_sym**2

    Psi_anti = (term1 - term2) / np.sqrt(2)
    prob_anti = Psi_anti**2

    P_total = 0.25 * prob_sym + 0.75 * prob_anti
    return P_total

L = 1.0
V0 = 250.0

limit_vis = 1.2
puntos = 100
x_vals = np.linspace(-limit_vis, limit_vis, puntos)
X1, X2 = np.meshgrid(x_vals, x_vals)

energias = encontrar_energias(L, V0)
psis_1d = generar_psis(x_vals, energias, L, V0)

print(f"Número de estados ligados encontrados: {len(energias)}")
if len(energias) < 5:
    print("¡ERROR! V0 es muy bajo, no existe el estado n=5.")

Z_a = get_P_spin_finite(1, 2, psis_1d)

Z_b = get_P_spin_finite(1, 5, psis_1d)

def dibujar_contornos(ax, Z_data, L, x_vec):
    idx_pos = (np.abs(x_vec - L/2)).argmin()
    idx_neg = (np.abs(x_vec + L/2)).argmin()
    val_p, val_n = x_vec[idx_pos], x_vec[idx_neg]
    style = {'color': 'black', 'linewidth': 3, 'linestyle': ':', 'zorder': 10}

    ax.plot(np.full_like(x_vec, val_p), x_vec, Z_data[:, idx_pos], **style, label='Borde del pozo')
    ax.plot(np.full_like(x_vec, val_n), x_vec, Z_data[:, idx_neg], **style)
    ax.plot(x_vec, np.full_like(x_vec, val_p), Z_data[idx_pos, :], **style)
    ax.plot(x_vec, np.full_like(x_vec, val_n), Z_data[idx_neg, :], **style)

    k = L/2
    ax.plot([-k, -k, k, k, -k], [-k, k, k, -k, -k], [0,0,0,0,0], 'k:', lw=2, zorder=2)

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X1, X2, Z_a, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=4, cstride=4, alpha=0.7, zorder=1)
dibujar_contornos(ax1, Z_a, L, x_vals)

ax1.set_title('(a) Fermiones con Espín (Pozo Finito)\n$n_1=1, n_2=2$')
ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.set_zlabel('$P$')
ax1.set_xlim(-limit_vis, limit_vis)
ax1.set_ylim(-limit_vis, limit_vis)
ax1.view_init(elev=35, azim=45)
ax1.legend()

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X1, X2, Z_b, cmap='Wistia',
                         edgecolor='black', linewidth=0.3, rstride=4, cstride=4, alpha=0.7, zorder=1)
dibujar_contornos(ax2, Z_b, L, x_vals)

ax2.set_title('(b) Fermiones con Espín (Pozo Finito)\n$n_1=1, n_2=5$')
ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.set_zlabel('$P$')
ax2.set_xlim(-limit_vis, limit_vis)
ax2.set_ylim(-limit_vis, limit_vis)
ax2.view_init(elev=35, azim=45)
ax2.legend()

fig.suptitle('Fermiones con espín (potencial finito)', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

```python
#@title Análisis del Límite Termodinámico: Densidad de Estados vs Ancho del Pozo

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

# Parámetros físicos (unidades adimensionales)
hbar = 1.0
m = 1.0

# Diferentes anchos del pozo para analizar
L_values = [1.0, 2.0, 5.0, 10.0, 20.0, 50.0]
n_max = 100  # Número máximo de estados a considerar
E_max = 200  # Energía máxima para análisis

# Colores para los diferentes anchos
colors = plt.cm.viridis(np.linspace(0, 1, len(L_values)))

# Crear figura con múltiples subplots
fig = plt.figure(figsize=(16, 12))
gs = GridSpec(3, 2, figure=fig, hspace=0.35, wspace=0.3)

# --- SUBPLOT 1: Espectro de Energías ---
ax1 = fig.add_subplot(gs[0, :])
for idx, L in enumerate(L_values):
    energias = [(np.pi**2 * hbar**2 * n**2) / (2 * m * L**2) for n in range(1, n_max + 1)]
    energias_filtradas = [E for E in energias if E <= E_max]
    n_estados = len(energias_filtradas)
    
    # Graficar como líneas horizontales
    for E in energias_filtradas:
        ax1.hlines(E, idx - 0.3, idx + 0.3, colors=colors[idx], linewidth=2, alpha=0.7)

ax1.set_xlabel('Ancho del Pozo L', fontsize=12, fontweight='bold')
ax1.set_ylabel('Energía $E_n$', fontsize=12, fontweight='bold')
ax1.set_title('(a) Espectro de Energías para Diferentes Anchos de Pozo', fontsize=14, fontweight='bold')
ax1.set_xticks(range(len(L_values)))
ax1.set_xticklabels([f'L={L}' for L in L_values])
ax1.set_ylim(0, E_max)
ax1.grid(True, alpha=0.3)

# --- SUBPLOT 2: Número Acumulado de Estados ---
ax2 = fig.add_subplot(gs[1, 0])
for idx, L in enumerate(L_values):
    energias = [(np.pi**2 * hbar**2 * n**2) / (2 * m * L**2) for n in range(1, n_max + 1)]
    energias_filtradas = sorted([E for E in energias if E <= E_max])
    
    # Número acumulado de estados
    N_acumulado = list(range(1, len(energias_filtradas) + 1))
    
    ax2.plot(energias_filtradas, N_acumulado, 'o-', color=colors[idx], 
             label=f'L={L}', linewidth=2, markersize=4, alpha=0.7)

# Curva teórica en el límite continuo: N(E) = L * sqrt(2mE) / (pi * hbar)
E_continuo = np.linspace(0.1, E_max, 500)
for idx, L in enumerate([L_values[-1]]):  # Solo para L más grande
    N_teorico = L * np.sqrt(2 * m * E_continuo) / (np.pi * hbar)
    ax2.plot(E_continuo, N_teorico, '--', color='red', linewidth=3, 
             label=f'Límite continuo (L={L})', alpha=0.8)

ax2.set_xlabel('Energía $E$', fontsize=12, fontweight='bold')
ax2.set_ylabel('Número Acumulado de Estados $N(E)$', fontsize=12, fontweight='bold')
ax2.set_title('(b) Número Acumulado de Estados', fontsize=14, fontweight='bold')
ax2.legend(fontsize=9, loc='upper left')
ax2.grid(True, alpha=0.3)

# --- SUBPLOT 3: Densidad de Estados (Histograma) ---
ax3 = fig.add_subplot(gs[1, 1])

# Bins de energía para calcular la densidad
E_bins = np.linspace(0, E_max, 50)
bin_width = E_bins[1] - E_bins[0]

for idx, L in enumerate(L_values[::2]):  # Cada 2 valores para no saturar
    energias = [(np.pi**2 * hbar**2 * n**2) / (2 * m * L**2) for n in range(1, n_max + 1)]
    energias_filtradas = [E for E in energias if E <= E_max]
    
    # Histograma de densidad
    counts, bins = np.histogram(energias_filtradas, bins=E_bins)
    densidad = counts / bin_width
    
    ax3.plot(bins[:-1] + bin_width/2, densidad, 'o-', color=colors[idx*2], 
             label=f'L={L}', linewidth=2, markersize=5, alpha=0.7)

# Densidad teórica en el límite continuo: ρ(E) = L/(π*hbar) * sqrt(2m/E)
E_teorico = np.linspace(1, E_max, 500)
for L in [L_values[-1]]:
    rho_teorico = (L / (np.pi * hbar)) * np.sqrt(2 * m / E_teorico)
    ax3.plot(E_teorico, rho_teorico, '--', color='red', linewidth=3, 
             label=f'$\\rho(E)$ teórica (L={L})', alpha=0.8)

ax3.set_xlabel('Energía $E$', fontsize=12, fontweight='bold')
ax3.set_ylabel('Densidad de Estados $\\rho(E)$', fontsize=12, fontweight='bold')
ax3.set_title('(c) Densidad de Estados (Estados por Unidad de Energía)', fontsize=14, fontweight='bold')
ax3.legend(fontsize=9, loc='upper right')
ax3.set_ylim(0, None)
ax3.grid(True, alpha=0.3)

# --- SUBPLOT 4: Separación entre Niveles ---
ax4 = fig.add_subplot(gs[2, 0])

for idx, L in enumerate(L_values):
    energias = [(np.pi**2 * hbar**2 * n**2) / (2 * m * L**2) for n in range(1, min(50, n_max) + 1)]
    
    # Calcular separación entre niveles consecutivos
    delta_E = [energias[i+1] - energias[i] for i in range(len(energias)-1)]
    E_medio = [(energias[i+1] + energias[i])/2 for i in range(len(energias)-1)]
    
    ax4.plot(E_medio, delta_E, 'o-', color=colors[idx], 
             label=f'L={L}', linewidth=2, markersize=4, alpha=0.7)

ax4.set_xlabel('Energía Media $E$', fontsize=12, fontweight='bold')
ax4.set_ylabel('Separación $\\Delta E = E_{n+1} - E_n$', fontsize=12, fontweight='bold')
ax4.set_title('(d) Separación entre Niveles Consecutivos', fontsize=14, fontweight='bold')
ax4.legend(fontsize=9, loc='upper right')
ax4.grid(True, alpha=0.3)
ax4.set_yscale('log')

# --- SUBPLOT 5: Escalamiento con L ---
ax5 = fig.add_subplot(gs[2, 1])

# Analizar el número de estados por debajo de una energía fija
E_fijo = 100.0
N_estados_list = []

L_scan = np.linspace(1, 100, 50)
for L in L_scan:
    count = 0
    n = 1
    while True:
        E = (np.pi**2 * hbar**2 * n**2) / (2 * m * L**2)
        if E > E_fijo:
            break
        count += 1
        n += 1
    N_estados_list.append(count)

ax5.plot(L_scan, N_estados_list, 'o-', color='darkblue', 
         linewidth=2, markersize=4, label=f'$N(E < {E_fijo})$')

# Teoría: N ~ L * sqrt(E)
N_teorico_scan = L_scan * np.sqrt(2 * m * E_fijo) / (np.pi * hbar)
ax5.plot(L_scan, N_teorico_scan, '--', color='red', linewidth=3, 
         label='Límite continuo: $N \\propto L\\sqrt{E}$', alpha=0.8)

ax5.set_xlabel('Ancho del Pozo $L$', fontsize=12, fontweight='bold')
ax5.set_ylabel(f'Número de Estados con $E < {E_fijo}$', fontsize=12, fontweight='bold')
ax5.set_title('(e) Escalamiento del Número de Estados con $L$', fontsize=14, fontweight='bold')
ax5.legend(fontsize=10, loc='upper left')
ax5.grid(True, alpha=0.3)

# Título general
fig.suptitle('Límite Termodinámico: Transición de Espectro Discreto a Continuo', 
             fontsize=16, fontweight='bold', y=0.995)

plt.show()

# --- ANÁLISIS CUANTITATIVO ---
print("=" * 70)
print("ANÁLISIS DEL LÍMITE TERMODINÁMICO")
print("=" * 70)
print(f"\nEvaluando para E < {E_max}:\n")
print(f"{'Ancho L':<10} {'N° Estados':<12} {'Densidad Media':<18} {'ΔE promedio':<15}")
print("-" * 70)

for L in L_values:
    energias = [(np.pi**2 * hbar**2 * n**2) / (2 * m * L**2) for n in range(1, n_max + 1)]
    energias_filtradas = [E for E in energias if E <= E_max]
    n_estados = len(energias_filtradas)
    
    if n_estados > 1:
        densidad_media = n_estados / E_max
        delta_E_list = [energias_filtradas[i+1] - energias_filtradas[i] 
                        for i in range(len(energias_filtradas)-1)]
        delta_E_promedio = np.mean(delta_E_list)
        
        print(f"{L:<10.1f} {n_estados:<12} {densidad_media:<18.4f} {delta_E_promedio:<15.6f}")

print("\n" + "=" * 70)
print("OBSERVACIONES:")
print("=" * 70)
print("1. A mayor L, mayor densidad de estados → Transición a continuo")
print("2. ΔE disminuye como 1/L² → Niveles más cercanos")
print("3. N(E) escala linealmente con L en el límite termodinámico")
print("4. ρ(E) ∝ L/√E → Singularidad en E=0 (característica 1D)")
print("=" * 70)
```

## Análisis del Límite Termodinámico: Densidad de Estados

En el límite termodinámico, cuando el ancho del pozo $L \rightarrow \infty$, se espera que el espectro discreto de energías se vuelva cada vez más denso, acercándose a un continuo. Este análisis es fundamental para entender la transición entre sistemas cuánticos discretos y el comportamiento clásico.

La densidad de estados $\rho(E)$ se define como el número de estados por unidad de energía. Para un pozo cuadrado infinito, las energías son:

$$E_n = \frac{h^2 n^2}{8mL^2} = \frac{\pi^2 \hbar^2 n^2}{2mL^2}$$

A medida que $L$ aumenta:
- Las energías se comprimen hacia valores más bajos
- La separación entre niveles consecutivos $\Delta E = E_{n+1} - E_n$ disminuye
- La densidad de estados $\rho(E)$ aumenta

En el límite $L \rightarrow \infty$, se recupera el resultado clásico:

$$\rho(E) = \frac{dN}{dE} = \frac{L}{\pi\hbar}\sqrt{\frac{2m}{E}}$$

El problema de la partícula confinada en una caja de potencial infinito proporciona una visión general de cómo se afrontan, de forma genérica, diversos problemas en mecánica cuántica. Esto se debe a que las condiciones de contorno imponen soluciones sinusoidales y energías discretas, las cuales son relativamente sencillas de determinar. En este trabajo se buscó caracterizar el sistema en términos de la posición a partir de la densidad de probabilidad, incorporando además una característica que usualmente se omite: la temperatura.

Al someter el sistema a un reservorio externo a temperatura $T$, se observó la influencia de este factor sobre los resultados. En particular, la densidad de probabilidad térmica mantiene una forma simétrica y, al aumentar la temperatura, tiende a volverse más uniforme (con una apariencia más “cuadrada”), pues la restricción asociada a las condiciones de borde afecta cada vez más el comportamiento del sistema.

Para el caso de dos partículas, el análisis mostró con claridad la diferencia entre fermiones y bosones, ya que sus funciones de onda deben ser antisimétrica y simétrica, respectivamente. Adicionalmente, se observó que la temperatura aumenta la probabilidad de encontrar las partículas cerca del borde debido a la aleatoriedad que introduce el reservorio; en bosones, por ejemplo, la cúspide característica tiende a pasar de una forma más puntiaguda a una más achatada, redistribuyéndose la probabilidad de manera más uniforme.

También se evidenció que, a partir de cierto número de términos en la sumatoria, los términos superiores no aportan de forma apreciable a la densidad de probabilidad, lo cual justifica el truncamiento numérico. Asimismo, la parte temporal no aparece en la densidad de probabilidad, dado que su contribución se manifiesta como una fase global que se elimina al calcular $|\Psi|^2$. Finalmente, al incluir el espín, la densidad de probabilidad requiere considerar un promedio sobre las contribuciones simétricas y antisimétricas que pueden formarse entre las partículas del sistema.
